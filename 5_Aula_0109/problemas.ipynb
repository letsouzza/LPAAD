```python
def texto_colorido(texto, cor, estilo):
    cores = {
        "preto": "30", 
        "vermelho": "31", 
        "verde": "32", 
        "amarelo": "33"}
    estilizacao = {"nenhum": "0", "negrito": "1", "italico": "3", "sublinhado": "4"}

    return f"\033[{estilizacao[estilo]}m\033[{cores[cor]}m{texto}\033[0m"
```

```python
print(texto_colorido("ola", "verde", "negrito"))
print("hello")
```

### Medidas Centrais 

```python
def calcular_media(lista):
    media = sum(lista)/ len(lista)
    return media
```

```python
def calcular_mediana(lista):
    quantidade = len(lista)
    lista_ordenada = sorted(lista)
    meio = quantidade // 2

    if quantidade % 2 == 0:
        return((lista_ordenada[meio] + lista_ordenada[meio -1]) / 2)
    else:
        return lista_ordenada[meio]
```

```python
print(calcular_mediana([23,14,31,17,20]))
print(calcular_mediana([75,1,4,3,19,50,64,31]))
```

```python
def calcular_moda(lista):

    moda = None
    frequencia = 0

    for valor in lista:
        if lista.count(valor) > frequencia:
            frequencia = lista.count(valor)
            moda = valor
            
    return moda
```

```python
conjunto_a = [ 0, 0, 30, 40]
conjunto_b = [10, 10, 20, 20, 30, 40]
conjunto_c = [10, 10, 20, 20, 30, 30, 40]
conjunto_d = [10, 20, 30, 40]

print(calcular_moda(conjunto_a))
print(calcular_moda(conjunto_b))
print(calcular_moda(conjunto_c))
print(calcular_moda(conjunto_d))
```

```python
def calcular_modas(valores):
    modas = []
    maior_frequencia = 0

    for valor in valores:
        frequencia = valores.count(valor)

        if frequencia > maior_frequencia:
            maior_frequencia = frequencia
            modas = [valor]

        elif frequencia == maior_frequencia and valor not in modas:
            modas.append(valor)

    if maior_frequencia == 1:
        return "Amodal", []
    elif len(modas) == 1:
        return "Unimodal", modas
    elif len(modas) == 2:
        return "Bimodal", modas
    else:
        return "Multimodal", modas
```

```python
conjunto_a = [ 0, 0, 30, 40]
conjunto_b = [10, 10, 20, 20, 30, 40]
conjunto_c = [10, 10, 20, 20, 30, 30, 40]
conjunto_d = [10, 20, 30, 40]

print(calcular_modas(conjunto_a))
print(calcular_modas(conjunto_b))
print(calcular_modas(conjunto_c))
print(calcular_modas(conjunto_d))
```
