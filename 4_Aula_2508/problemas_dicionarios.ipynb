#### Controle de estoque

```python
estoque = [18, 5, 0, 32, 4, 7, 0, 14, 3]
produtos_esgotados = 0
produtos_criticos = 0
produtos_disponiveis = 0

for valor in estoque:
    if valor == 0:
        produtos_esgotados += 1
    elif valor <= 5:
        produtos_criticos += 1
    else:
        produtos_disponiveis += 1

percentual_reposicao = ((produtos_criticos + produtos_esgotados) / len(estoque)) * 100

print(f"Relatório de estoque\nSua empresa tem {produtos_esgotados} produtos que estão esgotados"
        f"\nPossui {produtos_criticos} produtos que o estoque está crítico (abaixo de 5)"
        f"\nE {produtos_disponiveis} produtos estão disponíveis"
        f"\n\nPercentual de reposição: {percentual_reposicao:.2f}%")
```

#### Desempenho de 4 unidades

```python
empresa = [
    [120, 135, 128, 141],
    [98, 105, 110, 103],
    [150, 145, 160, 155],
    [116, 161, 108, 121]
]
total_atentidemento = []
maior_valor = empresa[0][0]

print("--------Relatório de Desempenho--------")

for unidade in range(len(empresa)):
    # Realizando soma de atendimentos da unidade 
    total = sum(empresa[unidade])
    print(f"\tUnidade {unidade + 1}: Possui total de atendimentos de {total}")

    # Analisando a unidade com maior quantidade de atendimentos 
    total_atentidemento.append(total)
    maximo_atendimento = max(total_atentidemento)

    # Acessando os valores de cada unidade 
    abaixo_meta = 0
    for valor in range(len(empresa[unidade])):

        # Verificando se os valores estão abaixo da meta 
        if empresa[unidade][valor] < 110:
            abaixo_meta += 1

        # Encontrando maior valor da semana em todas as unidades
        if empresa[unidade][valor] > maior_valor:
            maior_valor = empresa[unidade][valor]
    print(f"\tNa unidade {unidade + 1}, houve {abaixo_meta} itens abaixo da meta de 110 atendimentos!\n")
    

print(f"A unidade {unidade} teve a maior quantidade de atendimento, sendo {maximo_atendimento}")
print(f"O maior valor identificado foi da unidade {unidade + 1} sendo no total de {maior_valor} atendimentos")
```

#### Monitoramento de entregas 

```python
entregas = [
    {"motorista": "Bruno", "distancia": 32, "tempo": 48, "status": "Entregue"},
    {"motorista": "João", "distancia": 45, "tempo": 72, "status": "Atrasada"},
    {"motorista": "Pedro", "distancia": 51, "tempo": 85, "status": "Atrasada"},
    {"motorista": "José", "distancia": 49, "tempo": 81, "status": "Atrasada"}
]
atrasadas = 0
tempo_maximo = entregas[0]["tempo"]

print("---------Entregas---------")
for entrega in entregas:

    if entrega["status"] == "Atrasada":
        atrasadas += 1
        for chave, valor in entrega.items():
            print(f"{chave}: {valor}")
        print("\n")
    if entrega["tempo"] > tempo_maximo:
        tempo_maximo = entrega["tempo"]
        motorista = entrega["motorista"]

print(f"Quantidade de entregas atrasadas é de {atrasadas}\nSeu percentual de atraso é de {((atrasadas / len(entregas)) * 100):.2f} %")
print(f"A entrega que demorou mais tempo foi do motorista {motorista} e demorou {tempo_maximo} min")
```
