```python
venda = {
    "produto": "Notebook",
    "valor": 4200,
    "estoque": 8
}

for valor in venda.keys():
    print(f"Chaves: {valor}")
print("")

for valor in venda.values():
    print(f"Valores: {valor}")
print("")

for valor in venda.items():
    print(f"Pares: {valor}")
print("")
```

```python
vendas = [
    {"produto": "Notebook", "preco": 4200, "estoque": 8},
    {"produto": "Mouse", "preco": 120, "estoque": 25},
    {"produto": "Monitor", "preco": 1350, "estoque": 4},
    {"produto": "Teclado", "preco": 280, "estoque": 15},
    {"produto": "Headset", "preco": 450, "estoque": 3},
]
reposicao = 0
itens_repor = []

print("---------Produtos---------")
for produtos in vendas:
    print(f"\nProduto: {produtos["produto"]} Preço: {produtos["preco"]} Estoque: {produtos["estoque"]}")

    if produtos["estoque"] < 10:
        print(f"ATENÇÃO! {produtos["produto"]} - Estoque: {produtos["estoque"]}")
        reposicao += 1
        itens_repor.append(produtos["produto"])

print(f"\nTem {reposicao} produtos que precisam de reposição, sendo eles: {itens_repor}")
```
