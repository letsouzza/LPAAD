### Principais exceções em Python

* `ValueError`-> Valor inadequado (int("Felipe"))
* `TypeError` -> Tipos incompatíveis ("10" + 5)
* `ZeroDivisiomError` -> Divisão por zero
* `IndexError`-> Posição errada em lista
* `KeyError` -> Chave inexistente em dicionário
* `FileNotFoundError` -> Arquivo não foi encontrado

```python
while True:
    try: 
        idade = int(input("Digite sua idade: "))
        break
    except ValueError:
        print("Por favor, digite um número inteiro, tente novamente!")
```

#### Bloco Else:
Executa o else apenas se entrar no try 

```python
# Bloco else
try: 
    idade = int(input("Digite sua idade: "))
except ValueError:
    print("Por favor, digite um número inteiro, tente novamente!")
else:
    print("Idade:", idade)
```

#### Bloco finally:
Ele executa independente se a operação dá certo ou não 

```python
# Bloco finally
try: 
    idade = int(input("Digite sua idade: "))
except ValueError:
    print("Por favor, digite um número inteiro, tente novamente!")
finally:
    print("Operação finalizada!")
```

### Prática 1: Operação Bancária

```python
def juros_simples_composto(id, capital, parcelas, taxa= 0.02):
    if id == 1:
        juros_simples = capital * parcelas * taxa
        mont_simples = capital + juros_simples
        return juros_simples, mont_simples
    elif id == 2:
        mont_composto = capital * (pow((taxa + 1), parcelas))
        juros_composto = mont_composto - capital
        return round(juros_composto, 2), round(mont_composto, 2)
    else:
        return "Não foi possível calcular os valores", False
```

```python
try:
    saque = float(input("Digite um valor para saque:"))
    parcela = int(input("Quantidade de parcelas (1 a 6):"))

    print("Digite um método de Juros\n1- Juros Simples\n2- Juros Composto")
    id_juros = int(input("Digite um método de juros:"))

    valor_juros, montante = juros_simples_composto(id_juros, saque, parcela)

    limite = 8000
    parcelas_validas = [i for i in range(1,7)]

    valor_parcela = montante / parcela 
    limite_disponivel = limite - montante 

    if saque < 100 or saque > 5000:
        print("ERRO: Digite um valor de saque entre 100 e 5000")
    elif limite_disponivel < 2500:
        print("ERRO: O valor do limite não deve ser menor que 2500")
    elif parcela not in parcelas_validas:
        print("ERRO: A quantidade de parcelas deve ser entre 1 e 6")
    else:
        if valor_parcela < 50:
            print("ERRO: O valor minímo da parcela é R$50,00")
        else:
            print("\nSaque:", saque)
            print("Limite inicial:", limite )
            print(f"Limite disponível após saque: {limite_disponivel:.2f}")
            print("Quantidade de parcelas:", parcela)
            print(f"Valor da parcela: {valor_parcela:.2f}")
            print("Valor juros:", valor_juros)
            print("Montante final com acréscimo de juros:", montante)

except ValueError:
    print("ERRO: Digite um valor válido!")
except ZeroDivisionError:
    print("ERRO: Digite a quantidade de parcelas entre 1 e 6")
finally:
    print("\nOperação finalizada")
```

### Validando funções com `*args`

```python
import statistics 
```

```python
def registrar_notas(aluno, *notas):

    print(50*"-")

    if not isinstance(aluno, str) or not aluno:
        print("Nome inválido!")
        return 

    if len(notas) < 1:
        print("É necessário pelo menos 1 nota")
        return

    for nota in notas:
        if not isinstance(nota, (int, float)):
            print("As notas devem ser valores numéricos")
            return

        if nota < 0 or nota > 10:
            print("As notas devem estar entre 0 e 10")
            return 

    print("Aluno:", aluno)
    print("Notas:", notas)
    print("Quantidade de notas:", len(notas))
    print(f"Média: {statistics.mean(notas):.2f}")
```

```python
registrar_notas("Felipe", 8.0, 7.5, 9.0)
registrar_notas("", 8.0, 7.5)
registrar_notas("Felipe")
registrar_notas("Felipe", "n")
```
