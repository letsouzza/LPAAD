### Datetime 

**Módulo fornece recursos para trabalhar com Data e Hora**
* `data = datetime(2026, 9, 30, 18, 45)`
    * day
    * month
    * year
    * hour
    * minute
* Data atual: `datetime.now()`



**Formatação da Data:**
* `strftime()`: Transforma data em string apenas para apresentar ``("%d/%m/%Y %H:%M")``
    * %d -> dia
    * %m -> mês
    * %Y -> ano
    * %H -> hora
    * %M -> minuto



**STRING para DATETIME:**
* `strptime(data_texto, "formatação")`: Converte de string para data


```python
from datetime import datetime
```

```python
def converter_str_data(data):
    data_convertida = datetime.strptime(data, "%d/%m/%Y")
    return data_convertida 
```

```python
data_pedido = "12/09/2026"
data_entrega = "18/09/2026"

data_pedido_convertida = converter_str_data(data_pedido)
data_entrega_convertida = converter_str_data(data_entrega)

print("Analisando conversão\n")
print(f"Data Pedido antes: {type(data_pedido)}\nData Pedido depois: {type(data_pedido_convertida)}\n")
print(f"Data Entrega antes: {type(data_pedido)}\nData Entrega depois: {type(data_pedido_convertida)}")
```

```python
# Calculando diferença entre dias
diferenca_dias = data_entrega_convertida - data_pedido_convertida
dias_entrega = diferenca_dias.days
print(f"Diferença de dias: {dias_entrega}")
```

```python
# Apresentando datas no formato dd/mm/yy
print(f"Data pedido: {data_pedido_convertida.strftime("%d/%m/%Y")}")
print(f"Data entrega: {data_entrega_convertida.strftime("%d/%m/%Y")}")
```

```python
# Verificando se ocorreu no período 
if dias_entrega <= 5:
    print("Entrega ocorreu em até 5 dias")
else:
    print("Entrega demorou mais de 5 dias")
```

### Enumerate

Contador de forma automatizada em uma iteração 
* `for indice, produto in enumerate(lista, start=1):`
    * start -> onde inicia a contagem do valor 

```python
vendedores = [
    {"nome": "Ana","vendas":12500},
    {"nome": "Carlos","vendas":9800},
    {"nome": "Marina","vendas":15700},
    {"nome": "Rafael","vendas":11300}
]

print("RELATÓRIO FUNCIONÁRIOS\n")

for i, vendedor in enumerate(vendedores, start=1):
    if vendedor['vendas'] >= 10000:
        print(f"{i}- Nome: {vendedor['nome']} | Vendas: {vendedor['vendas']}, Atingiu a META!")
    else:
        print(f"{i}- Nome: {vendedor['nome']} | Vendas: {vendedor['vendas']}, Meta NÃO atingida!")
        
```

### Args e Kwargs

* `*args`: permite que uma função receba uma quantidade variável de argumentos posicionais na chamada 
    * Ex: vou calcular a nota de alunos, mas não sei a quantidade de alunos 

* `**kwargs`: passar argumentos nomeados adicionais (par chave e valor -> cidade= "São Paulo")
    * Ex: passar dados de um cliente, como "dicionário"


```python
# Prática do Args 
def registrar_treinamento(tema, *participantes):

    print(f"O tema do treinamento é: {tema}")
    print(f"Há {len(participantes)} participante(s) neste treinamento")

    for numero, participante in enumerate(participantes, start=1):
        print(f"{numero}° Participante: {participante}")
```

```python
registrar_treinamento("SQL", "João", "Lapa", "Leticia", "Leo", "Ana")
```

```python
# Prática do Kwargs 
def registro_ocorrencia(codigo, *sistemas, **informacoes):
    print(f"Quantidade de sistemas afetados: {len(sistemas)}\n")
    print("Sistemas:")
    for i, sist in enumerate(sistemas, start=1):
        print(f"{i}- {sist}")
    print(f"\nInformações:")
    for chave, valor in informacoes.items():
        print(f"{chave}: {valor}")
```

```python
registro_ocorrencia(
    "INC-001",
    "Servidor Web",
    "Banco de Dados",
    "API",
    prioridade= "Alta",
    responsavel= "Lapa",
    status= "Em análise"
)
```
