## Problema Final- Processamento de Encomendas 

Uma empresa de logística realiza o transporte de encomendas para diferentes regiões do
Brasil. Cada solicitação pode conter um ou mais pacotes, com pesos distintos, e o custo do
transporte depende da região de destino, da modalidade escolhida e da contratação de seguro.

```python
def calcular_frete(cliente, *pesos, **informacoes):

    #----------------------------------------------
    # REGRAS DE NEGOCIO 
    #----------------------------------------------
    # Verificação do nome 
    if not isinstance(cliente, str) or not cliente.strip():
        print("ERRO: Nome inválido!")
        return 
    

    # Validações de peso 
    if len(pesos) < 1:
        print("ERRO: Insira ao menos um peso")
        return
    
    for peso in pesos:
        total_pesos= sum(pesos)
        if isinstance(peso, (bool)):
            print("ERRO: Os pesos devem ser valores numéricos")
            return

        if not isinstance(peso, (int, float)):
            print("ERRO: Os pesos devem ser valores numéricos")
            return

        if peso <= 0 or peso > 30:
            print("ERRO: Os pesos devem estar entre 1 e 30")
            return 
        
        if total_pesos > 100:
            print("ERRO: Não é permitido que o peso total ultrapasse 100 kg")
            return


    # Validação de Região 
    if "regiao" not in informacoes:
        print("ERRO: É necessário que adicione a região")
        return

    tarifas_regiao = {
        "sudeste": 5,
        "sul": 6,
        "centro-oeste": 7,
        "nordeste": 8,
        "norte": 10
    }
    if informacoes["regiao"].lower() not in tarifas_regiao.keys():
        print("ERRO: Digite uma região entre as permitidas: sul, sudeste, centro-oeste, norte e nordeste")
        return

    # Validação de Modalidade 
    if "modalidade" not in informacoes:
        print("ERRO: É necessário que adicione a modalidade")
        return
    
    modalidades = ["economica", "expressa"]
    if informacoes["modalidade"].lower() not in modalidades:
        print("ERRO: A modalidade deve ser Econômica ou Expressa")
        return

    # Validação de seguro
    if "seguro" in informacoes:
        if not isinstance(informacoes["seguro"], bool):
            print("ERRO: Insira um seguro válido, devendo ser booleano!")
            return
    else:
        informacoes["seguro"]= False
    #----------------------------------------------
    

    #----------------------------------------------
    # CÁLCULO FRETE 
    #----------------------------------------------
    for chave, valor in tarifas_regiao.items():
        if informacoes["regiao"].lower() == chave:
            frete_base = total_pesos * valor 


    #----------------------------------------------
    # CÁLCULO MODALIDADE 
    #----------------------------------------------
    if informacoes["modalidade"].lower() == "expressa":
        frete_transporte = frete_base * 1.40
    else:
        frete_transporte = frete_base 


    #----------------------------------------------
    # CÁLCULO SEGURO
    #----------------------------------------------
    if "seguro":
        frete_final = frete_transporte * 1.05
    else:
        frete_final = frete_transporte 

    valor_seguro = frete_final - frete_transporte


    #----------------------------------------------
    # RESPOSTA
    #----------------------------------------------
    resultado = {
        "cliente": cliente,
        "peso_total": total_pesos,
        "frete_base": frete_base,
        "frete_modalidade": frete_transporte,
        "seguro": valor_seguro,
        "frete_final": frete_final,
        "situacao": "Encomenda Aprovada!"
    }

    return resultado
```

```python
def texto_colorido(texto, cor, estilo):
    cores = {
        "preto": "30", 
        "vermelho": "31", 
        "verde": "32", 
        "amarelo": "33"}
    estilizacao = {"nenhum": "0", "negrito": "1", "italico": "3", "sublinhado": "4"}

    return f"\033[{estilizacao[estilo]}m\033[{cores[cor]}m{texto}\033[0m"
```

```python
print("Relatório de resultados\n")
try:
    resultado = calcular_frete(
        "Felipe", 
        5, 10, 3,
        regiao="Sudeste",
        modalidade="Expressa",
        seguro=True
    )
    
    print(f"Cliente: {resultado['cliente']}")
    print(f"Peso total: {resultado['peso_total']} kg")
    print(f"Frete-base: R$ {resultado['frete_base']:.2f}".replace(".", ","))
    print(f"Frete com modalidade: R$ {resultado['frete_modalidade']:.2f}".replace(".", ","))
    print(f"Seguro: R$ {resultado['seguro']:.2f}".replace(".", ","))
    print(f"Frete final: R$ {resultado['frete_final']:.2f}".replace(".", ","))
    print(texto_colorido(resultado["situacao"], "verde", "negrito"))
    
except (TypeError, ValueError) as erro:
    print(f"Erro inesperado ao processar o teste: {erro}")
```
