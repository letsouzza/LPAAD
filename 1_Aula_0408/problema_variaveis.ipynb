#### 1. Desenvolvendo Calculadora básica 

```python
print("Olá, essa é a Best Cálculo a calculadora que realiza sete operações de uma vez!\n")
valor_1 = float(input("Digite o primeiro valor: "))
valor_2 = float(input("Digite o segundo valor: "))

soma = valor_1 + valor_2
diferenca = valor_1 - valor_2
produto = valor_1 * valor_2
quociente = valor_1 / valor_2
resto = valor_1 % valor_2
div_inteira = valor_1 // valor_2
potencia = valor_1 ** valor_2
raiz = valor_1 ** (1/2)

print(f"Resultados: \n\nSoma ->\t {soma}\nSubtração ->\t {diferenca}\nMultiplicação ->\t {produto}\nDivisão ->\t {quociente}\nResto da Divisão ->\t {resto}\nDivisão Inteira ->\t {div_inteira}\nPotência ->\t {potencia:.1f}\nRaiz ->\t {raiz:.1f}")
```
