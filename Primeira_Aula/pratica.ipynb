#### Tipos de Dados 

```python
nome_animal = input("Digite o nome do seu animal de estimação (ou crie um nome se não possuir): ")
num_preferido = input("Digite o seu número inteiro preferido: ")
altura = input("Digite sua altura: ")

print(type(nome_animal))
print(type(num_preferido))
print(type(altura))
```

Essa forma acima está incorreta, pois o computador compreende que todasas variáveis são string, mesmo tendo a presença de inteiros e float, por isso é importante ao pedir o input colocar o tipo adequado 

```python
nome_animal2 = input("Digite o nome do seu animal de estimação (ou crie um nome se não possuir): ")
num_preferido2 = int(input("Digite o seu número inteiro preferido: "))
altura2 = float(input("Digite sua altura: "))

print(type(nome_animal2))
print(type(num_preferido2))
print(type(altura2))
```

#### Caracteres Especiais no Print

* `**\n**` — Pula para a próxima linha (quebra de linha).

* `**\t**` — Adiciona um espaço de tabulação (tab).

* `\\` — Imprime uma barra invertida literal no texto.

* `**\' ou \"**` — Permite mostrar aspas simples ou duplas dentro de uma string que usa o mesmo tipo de aspa

* `**variavel:.2f**` - Formatar as casas decimais de um número no f string 

```python
salario = float(input("Digite seu salário atual:"))
salario_novo = salario * 1.15 

print(f"Olá {nome} \nSeu salário era de R${salario} ele sofreu um aumento de 15%.\nAgora seu salário atual é {salario_novo:.2f}")
```
