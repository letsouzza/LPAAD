## For\While- Aula 18/08

* **FOR** -> percorrer e analisar elemento por elemento (conheço o fim)
* **WHILE** -> Repetir ação até condição for verdadeira (não conhece o fim)

#### Variáveis Auxiliares 
* Contador: Inicia no zero e acrescenta em inteiros 
* Somatório: Inicia ao zero e soma em cada interação
* Produto: Inicia no um e multiplica em cada interação 

#### Contadores de defeitos 

```python
print("hello world")
```

```python
defeitos = [2, 0, 4, 1, 0, 3, 0, 5]
contador = 0

for defeito in defeitos:
    if defeito > 0:
        contador += 1
print("Quantidade de defeitos:", contador)
```

#### Tempo máximo

```python
tempos = [32, 48, 27, 55, 41, 29]
contador = 0
soma = 0
atrasadas = 0

print("Contabilizando tempos de entrega:")
for tempo in tempos:
    contador += 1
    if tempo > 40:
        condicao = "Atrasado"
        atrasadas += 1
    else:
        condicao = "Entregue no prazo"
    print(f"Índice: {contador}\nTempo: {tempo}\nCondição: {condicao}\n")
    soma += tempo

porcentagem_atraso = (atrasadas/contador) * 100

print("Tempo total das entregas:", soma)
print(f"Tempo médio das entregas {soma/contador:.2f}")
print(f"Quantidade de entregas atrasadas: {atrasadas}\n Porcentagem de atraso: {porcentagem_atraso}%")
```

#### Controle de Produção

```python
producoes = []
peca = int(input("Digite a quantidade de peças produzidas por turno (informe valores entre 300 e 800 peças, ou -1 caso finalize a qauntidade): "))
total = 0
quant_turnos_meta = 0

while peca != -1:
    if peca < 300 or peca > 800:
        print("ERRO! Só aceitamos valor entre 300 e 800, digite outro valor")
        peca = int(input("Digite a quantidade de peças produzidas por turno (informe valores entre 300 e 800 peças, ou -1 caso finalize a qauntidade): "))
    else:
        producoes.append(peca)
        peca = int(input("Digite a quantidade de peças produzidas por turno (informe valores entre 300 e 800 peças, ou -1 caso finalize a qauntidade): "))

for valor in producoes:
    if valor >= 500:
        condicao = "Meta atingida!"
        quant_turnos_meta += 1
    elif valor >= 400:
        condicao = "Abaixo da meta"
    else:
        condicao = "Crítico !!!!!"

    total += valor 

print(producoes)
print(f"\nRelatório final de produtividade\n"
      f"A produção total foi {total} isso em {len(producoes)} turnos.\n"
      f"A produção média foi {total/len(producoes):.2f}\n"
      f"Dentre os valores em cada turno, {quant_turnos_meta} atigiram a meta. Cerca de {(quant_turnos_meta/len(producoes)) * 100:.2f}% atingiram a meta!")
```
