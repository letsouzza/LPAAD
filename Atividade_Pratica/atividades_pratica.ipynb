### *Prática 1: Sistema de Acompanhamento de Chamados*

#### Parte 1: Processamento dos chamados

```python
def converter_str_data(data):
    data_convertida = datetime.strptime(data, "%d/%m/%Y")
    return data_convertida
```

```python
chamados = [
  {
  "codigo": "CH001",
  "responsavel": "Maria",
  "abertura": "15/09/2026",
  "fechamento": "17/09/2026",
  "prioridade": "Alta"
  },
  {
  "codigo": "CH002",
  "responsavel": "Carlos",
  "abertura": "16/09/2026",
  "fechamento": "22/09/2026",
  "prioridade": "Média"
  },
  {
  "codigo": "CH003",
  "responsavel": "Marina",
  "abertura": "18/09/2026",
  "fechamento": "19/09/2026",
  "prioridade": "Alta"
  },
  {
  "codigo": "CH004",
  "responsavel": "Rafael",
  "abertura": "20/09/2026",
  "fechamento": "25/09/2026",
  "prioridade": "Baixa"
  },
  {
  "codigo": "CH005",
  "responsavel": "Maria",
  "abertura": "21/09/2026",
  "fechamento": "23/09/2026",
  "prioridade": "Média"
  }
]

for i, chamado in enumerate(chamados, start=1):
    #Convertendo datas
    chamado['abertura'] = converter_str_data(chamado['abertura'])
    chamado['fechamento'] = converter_str_data(chamado['fechamento'])

    # Calculando diferença entre dias
    diferenca_dias = chamado['fechamento'] - chamado['abertura']
    chamado["dias_entrega"] = diferenca_dias.days

    if chamado['dias_entrega'] <= 3:
        chamado['prazo'] = "Dentro do Prazo!"
        print(f"{i}- {chamado['codigo']} | {chamado['responsavel']} | {chamado['prioridade']} | {chamado['dias_entrega']} dias | {chamado['prazo']}")
    else:
        chamado['prazo'] = "Fora do Prazo!"
        print(f"{i}- {chamado['codigo']} | {chamado['responsavel']} | {chamado['prioridade']} | {chamado['dias_entrega']} dias | {chamado['prazo']}")
```

#### Parte 2: Resumo dos atendimentos


```python
qnt_chamados = len(chamados)
chamados_dentro_prazo = 0
chamados_fora_prazo = 0
max_dias_entrega = chamados[0]['dias_entrega']
codigo_maior_atendimento = None
prioridade_alta= 0
prioridade_media= 0
prioridade_baixa = 0

for chamado in chamados:
    # Quantidade chamados dentro e fora do prazo
    if chamado['prazo'] == "Dentro do Prazo!":
        chamados_dentro_prazo += 1
    else:
        chamados_fora_prazo += 1

    # Encontrar o maior tempo de atendimento
    if chamado['dias_entrega'] > max_dias_entrega:
        max_dias_entrega = chamado['dias_entrega']
        codigo_maior_atendimento = chamado['codigo']

    # Prioridade quantidades
    if chamado['prioridade'] == "Alta":
        prioridade_alta += 1
    elif chamado['prioridade'] == "Média":
        prioridade_media += 1
    elif chamado['prioridade'] == "Baixa":
        prioridade_baixa += 1

print(f'Quantidade total de chamados: {qnt_chamados}\nQuantidade chamados dentro do prazo: {chamados_dentro_prazo}\nQuantidade chamados fora do prazo: {chamados_fora_prazo}\nMaior tempo de atendimento: {max_dias_entrega} dias no chamado {codigo_maior_atendimento}\n')
print(f'Quantidade chamados com prioridade Alta: {prioridade_alta}\nQuantidade chamados com prioridade Média: {prioridade_media}\nQuantidade chamados com prioridade Baixa: {prioridade_baixa}')
```

#### Parte 3: Organização por prioridade

```python
chamados_por_prioridade = {
  "Alta": [],
  "Média": [],
  "Baixa": []
}

for chamado in chamados:
    if chamado['prioridade'] == "Alta":
        chamados_por_prioridade["Alta"].append(chamado['codigo'])
    elif chamado['prioridade'] == "Média":
        chamados_por_prioridade["Média"].append(chamado['codigo'])
    elif chamado['prioridade'] == "Baixa":
        chamados_por_prioridade["Baixa"].append(chamado['codigo'])

print(chamados_por_prioridade)
```

#### Desafio

```python
fora_do_prazo = []

for chamado in chamados:
    if chamado['prazo'] == "Fora do Prazo!":
        fora_do_prazo.append({
            "codigo": chamado['codigo'],
            "responsavel": chamado['responsavel'],
            "tempo": chamado['dias_entrega']})
print("Chamados fora do Prazo")
print(fora_do_prazo)
```

#### Parte 2: Análise estatística dos valores

```python
def calcular_modas(valores):
    modas = []
    maior_frequencia = 0

    for valor in valores:
        frequencia = valores.count(valor)

        if frequencia > maior_frequencia:
            maior_frequencia = frequencia
            modas = [valor]

        elif frequencia == maior_frequencia and valor not in modas:
            modas.append(valor)

    if maior_frequencia == 1:
        return "Amodal", []
    elif len(modas) == 1:
        return "Unimodal", modas
    elif len(modas) == 2:
        return "Bimodal", modas
    else:
        return "Multimodal", modas
```

```python
def calcular_variancia(lista):
    qnt_lista = len(lista)
    media = statistics.mean(lista)
    somatoria = 0

    for elemento in lista:
        diferenca_quadrado = pow((elemento - media),2)
        somatoria += diferenca_quadrado

    variancia = somatoria / qnt_lista

    return variancia
```

```python
import math
def calcular_desvio_padrao(lista):
    variancia = calcular_variancia(lista)
    desvio_padrao = math.sqrt(variancia)

    return desvio_padrao
```

```python
def medidas_analise(lista):
    media = statistics.mean(lista)
    mediana = statistics.median(lista)
    moda = calcular_modas(lista)
    variancia = calcular_variancia(lista)
    desvio = calcular_desvio_padrao(lista)
    maximo = max(lista)
    minimo = min(lista)

    return media, mediana, moda, variancia, desvio, maximo, minimo
```

```python
from ast import mod
valores = []

for transacao in transacoes:
    valores.append(transacao['valor'])

media, mediana, moda, variancia, desvio, maximo, minimo = medidas_analise(valores)
print("Resultados Encontrados:")
print(f" Média: {media}\n Mediana: {mediana:.2f}\n Moda: {moda}\n Variância: {variancia:.2f}\n Desvio: {desvio:.2f}\n Máximo: {maximo}\n Mínimo: {minimo}")
```

#### Parte 3: Análise das transações

```python
for i, transacao in enumerate(transacoes, start=1):
  if transacao['valor'] > media:
    transacao['classificacao_media'] = "Acima da Média!"
  elif transacao['valor'] < media:
    transacao['classificacao_media'] = "Abaixo da Média!"
  else:
    chamado['prazo'] = "Na Média!"
  print(f"{i}- {transacao['codigo']} | {transacao['cliente']} | R$ {transacao['valor']:.2f} | {transacao['classificacao_media']}")
```

#### Parte 4: Resumo das transações


```python
qnt_transacoes = len(transacoes)
valor_total = sum(valores)
qnt_acima_media = 0
qnt_abaixo_media = 0
max_valor = transacoes[0]['valor']
min_valor = transacoes[0]['valor']
codigo_max_valor = None
codigo_min_valor = None

for transacao in transacoes:
    if transacao['valor'] > max_valor:
        max_valor = transacao['valor']
        codigo_max_valor = transacao['codigo']

    if transacao['valor'] < min_valor:
        min_valor = transacao['valor']
        codigo_min_valor = transacao['codigo']

    if transacao['classificacao_media'] == "Acima da Média!":
        qnt_acima_media += 1
    elif transacao['classificacao_media'] == "Abaixo da Média!":
        qnt_abaixo_media += 1

print(f'Quantidade total de transações: {qnt_transacoes}\nValor total transações: {valor_total}\nMaior valor: R$ {max_valor} na transação {codigo_max_valor}\nMenor valor: R$ {min_valor} na transação {codigo_min_valor}\n')
print(f'Quantidade transações acima da média: {qnt_acima_media}\nQuantidade transações abaixo da média: {qnt_abaixo_media}')
```
