### Criando Classe Filme

```python
class Filme:
    pass
```

```python
# Criando objetos
f1 = Filme()
f2 = Filme()
```

```python
# Adicionando atributos 
f1.nome = "Carros 1"
f1.genero = "Ação"
f1.produtor = "Pixar"
f1.duracao = 1200

f2.nome = "Vingadores: Ultimato"
f2.genero = "Ação"
f2.produtor = "Disney- Marvel"
f2.duracao = 2100
```

```python
# Exibir os objetos
print(f"Nome do filme: {f1.nome}")
print(f"Gênero: {f1.genero}")
print(f"Produtor: {f1.produtor}")
print(f"Duração em minutos: {f1.duracao}")

print("-"*50)

print(f"Nome do filme: {f2.nome}")
print(f"Gênero: {f2.genero}")
print(f"Produtor: {f2.produtor}")
print(f"Duração em minutos: {f2.duracao}")
```

```python
# Modificar o título
f1.nome = "Minha mãe é uma Peça"
f1.genero = "Comédia"
f1.produtor = "Midgal Filmes"
f1.duracao = 1224

print(f"Nome do filme: {f1.nome}")
print(f"Gênero: {f1.genero}")
print(f"Produtor: {f1.produtor}")
print(f"Duração em minutos: {f1.duracao}")
```
