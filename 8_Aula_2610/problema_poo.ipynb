### Criando Classe Jogos

```python
class Jogos:
    def __init__(self, titulo: str, genero: str, desenvolvedora: str, duracao_campanha: int, preco: float,nota: float):
        self.titulo = titulo
        self.genero = genero 
        self.desenvolvedora = desenvolvedora
        self.duracao_campanha = duracao_campanha
        self.preco = preco
        self.nota = nota

        # Atributos de controle
        self.reproducoes = 0
        self.tocando = False

    # Padronize os titulos das funções 
    # Evite colocar titulo com o mesmo titulo do atributo
    def mostrar_titulo(self):
        print(f"Titulo do jogo: {self.titulo}")

    def mostrar_genero(self):
        print(f"Genero(a): {self.genero}")

    def mostrar_desenvolvedora(self):
        print(f"Desenvolvedora: {self.desenvolvedora}")

    def mostrar_duracao_campanha(self):
        print(f"Duração: {self.duracao_campanha}")

    def mostrar_nota(self):
        print(f"nota: {self.nota}")

    def mostrar_preco(self):
        print(f"Preço: R$ {self.preco:.2f}")


    # Mostrar todas as definições bases
    def mostrar_detalhe(self):
        print("Dados do jogo:")
        self.mostrar_titulo()
        self.mostrar_genero()
        self.mostrar_desenvolvedora()
        self.mostrar_duracao_campanha()
        self.mostrar_nota()
        self.mostrar_preco()


    def registrar_conquista(self, nova_conquista: str):
        self.conquista = nova_conquista
        print(f"Conquista registrada: {self.conquista}")
    
    def mostar_conquista(self):
        if hasattr(self, 'conquista'):
            print(f"Conquista: {self.conquista}")
        else:
            print("Nenhuma conquista registrada.")
```

```python
j1 = Jogos("Minecraft", "Sandbox", "Mojang", 100, 99.90, 9.0)
j2 = Jogos("The Sims 4", "Simulação", "Maxis", 50, 0.00, 8.5)
j3 = Jogos("Stardew Valley", "RPG", "ConcernedApe", 60, 24.99, 9.5)

# Mostrar os detalhes de cada jogo
j1.mostrar_detalhe()
print("\n")

j2.mostrar_detalhe()
print("\n")

j3.mostrar_detalhe()
print("\n")

# Registrar uma conquista em dois jogos
j1.registrar_conquista("Construiu sua primeira casa")
j3.registrar_conquista("Completou o primeiro ano na fazenda")

print("\nVerificando as conquistas:")

# Verificar a conquista de todos os jogos
j1.mostar_conquista()
j2.mostar_conquista()
j3.mostar_conquista()
```
