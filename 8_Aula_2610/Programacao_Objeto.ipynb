### Utilizando `__init__`

`--init--` -> Método especial do Python executado automaticamente durante a inicialização de uma instância. Permite estabelecer valores iniciais dos atributos de um objeto

`self` -> É o parâmetro convencional utilizado para referenciar a própria instância dentro de um método

```python
class MusicaInit:
    def __init__(self, nome, artista):
        self.nome = nome
        self.artista = artista
```

```python
class Musica:
    def __init__(self, nome: str, cantor: str, genero: str, duracao: int, ano: int):
        self.nome = nome
        self.cantor = cantor 
        self.genero = genero
        self.duracao = duracao
        self.ano = ano

        # Atributos de controle
        self.reproducoes = 0
        self.tocando = False

    # Padronize os nomes das funções 
    # Evite colocar nome com o mesmo nome do atributo
    def mostrar_nome(self):
        print(f"Nome da música: {self.nome}")

    def mostrar_cantor(self):
        print(f"Cantor(a): {self.cantor}")

    def mostrar_genero(self):
        print(f"Gênero: {self.genero}")

    def mostrar_duracao(self):
        print(f"Duração: {self.duracao}")

    def mostrar_ano(self):
        print(f"Ano: {self.ano}")


    # Mostrar todas as definições bases
    def mostrar_detalhe(self):
        print("Faixa Completa")
        self.mostrar_nome()
        self.mostrar_cantor()
        self.mostrar_genero()
        self.mostrar_duracao()
        self.mostrar_ano()


    # Reproduzir música
    def reproduzir(self):
        if self.tocando == False:
            self.tocando = True 
            self.reproducoes += 1
            print(f"Reproduzindo: {self.nome}")
        else:
            print("Essa musica já está sendo reproduzida!")


    # Pausar música
    def pausar(self):
        if self.tocando:
            self.tocando = False 
            print("Música pausada!")
        else:
            print("Nenhuma música está sendo reproduzida!")
```

```python
m1 = Musica("Caneta Azul", "Manoel Gomes", "Brega", 73, 2019)

m1.mostrar_nome()
m1.mostrar_cantor()
m1.mostrar_genero()
m1.mostrar_duracao()
m1.mostrar_ano()

print("\n")

m1.mostrar_detalhe()

print("\n")

m1.reproduzir()
print("Quantidade de reproduções:", m1.reproducoes)
```

```python
m1.reproduzir()
print("Quantidade de reproduções:", m1.reproducoes)

m1.reproduzir()

print("\n")
m1.pausar()
print("\n")

m1.reproduzir()
print("Quantidade de reproduções:", m1.reproducoes)
```
