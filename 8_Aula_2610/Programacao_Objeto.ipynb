### Utilizando `__init__`

`--init--` -> Método especial do Python executado automaticamente durante a inicialização de uma instância. Permite estabelecer valores iniciais dos atributos de um objeto

`self` -> É o parâmetro convencional utilizado para referenciar a própria instância dentro de um método

```python
class MusicaInit:
    def __init__(self, nome, artista):
        self.nome = nome
        self.artista = artista
```

```python
class Musica:
    def __init__(self, titulo: str, cantor: str, genero: str, duracao: int, ano: int):

        # Validação 
        # Duração
        if type(duracao) is not int or duracao <=0:
            raise ValueError("O atributo duração deve ser um número inteiro positivo")

        # Ano
        if type(ano) is not int or ano <=0:
            raise ValueError("O atributo ano deve ser um número inteiro positivo")

        # Título
        if not isinstance(titulo, str) or not titulo.strip():
            raise ValueError("O título da música deve ser um texto não vazio")

        # Cantor
        if not isinstance(cantor, str) or not cantor.strip():
            raise ValueError("O cantor da música deve ser um texto não vazio")
        
        # Gênero
        if not isinstance(genero, str) or not genero.strip():
            raise ValueError("O gênero da música deve ser um texto não vazio")

        self._titulo = titulo.strip()
        self._cantor = cantor.strip()
        self._genero = genero.strip()
        self._duracao = duracao
        self._ano = ano

        # Atributos de controle
        self.reproducoes = 0
        self.tocando = False
        self.favorito = False

    #====================================================
    #
    # PROPRIEDADES INTERNAS
    #
    #====================================================
    @property
    def titulo(self):
        return self._titulo

    @property
    def cantor(self):
        return self._cantor

    @property
    def genero(self):
        return self._genero

    @property
    def duracao(self):
        return self._duracao

    @property
    def ano(self):
        return self._ano

    #====================================================
    #
    # MÉTODOS DE VISUALIZAÇÃO
    #
    #====================================================
    def mostrar_titulo(self):
        print(f"titulo da música: {self._titulo}")

    def mostrar_cantor(self):
        print(f"Cantor(a): {self._cantor}")

    def mostrar_genero(self):
        print(f"Gênero: {self._genero}")

    def mostrar_duracao(self):
        print(f"Duração: {self._duracao}")

    def mostrar_ano(self):
        print(f"Ano: {self._ano}")

    # Mostrar todas as definições bases
    def mostrar_detalhe(self):
        print("Faixa Completa")
        self.mostrar_titulo()
        self.mostrar_cantor()
        self.mostrar_genero()
        self.mostrar_duracao()
        self.mostrar_ano()


    #====================================================
    #
    # MÉTODOS DE CONTROLE
    #
    #====================================================
    # Reproduzir música
    def reproduzir(self):
        if self.tocando == False:
            self.tocando = True 
            self.reproducoes += 1
            print(f"Reproduzindo: {self._titulo}")
        else:
            print("Essa musica já está sendo reproduzida!")

    # Pausar música
    def pausar(self):
        if self.tocando:
            self.tocando = False 
            print("Música pausada!")
        else:
            print("Nenhuma música está sendo reproduzida!")

    # Favoritar música
    def favoritar(self):
        if self.favorito == False:
            self.favorito = True
            print("Música favoritada! ❤️")
        else:
            print("Música já favoritada")

    # Desfavoritar música
    def desfavoritar(self):
        if self.favorito:
            self.favorito = False
            print("Música desfavoritada!")
        else:
            print("Música não está favoritada")


    #====================================================
    #
    # MÉTODOS DE ATUALIZAÇÃO DE ATRIBUTOS
    #
    #====================================================
    # Atualizar titulo
    def alterar_titulo(self, novo_titulo: str):
        if not isinstance(novo_titulo, str) or not novo_titulo.strip():
            raise ValueError("O título da música deve ser um texto não vazio")
        
        self._titulo = novo_titulo.strip()
        print(f"Novo título da música: {self._titulo}")

    # Atualizar cantor
    def alterar_cantor(self, novo_cantor: str):
        if not isinstance(novo_cantor, str) or not novo_cantor.strip():
            raise ValueError("O cantor da música deve ser um texto não vazio")
        
        self._cantor = novo_cantor.strip()
        print(f"Novo cantor da música: {self._cantor}")

    # Atualizar genero
    def alterar_genero(self, novo_genero: str):
        if not isinstance(novo_genero, str) or not novo_genero.strip():
            raise ValueError("O gênero da música deve ser um texto não vazio")
        
        self._genero = novo_genero.strip()
        print(f"Novo gênero da música: {self._genero}")

    # Atualizar duracao
    def alterar_duracao(self, novo_duracao: int):
        if type(novo_duracao) is not int or novo_duracao <=0:
            raise ValueError("O atributo duração deve ser um número inteiro positivo")

        self._duracao = novo_duracao
        print(f"Nova duração da música: {self._duracao}")

    # Atualizar ano
    def alterar_ano(self, novo_ano: int):
        if type(novo_ano) is not int or novo_ano <=0:
            raise ValueError("O atributo ano deve ser um número inteiro positivo")

        self._ano = novo_ano
        print(f"Novo ano da música: {self._ano}")
```

```python
m1 = Musica("Caneta Azul", "Manoel Gomes", "Brega", 73, 2019)

m1.mostrar_titulo()
m1.mostrar_cantor()
m1.mostrar_genero()
m1.mostrar_duracao()
m1.mostrar_ano()

print("\n")

m1.mostrar_detalhe()

print("\n")

m1.reproduzir()
print("Quantidade de reproduções:", m1.reproducoes)
```

```python
m1.reproduzir()
print("Quantidade de reproduções:", m1.reproducoes)

m1.reproduzir()

print("\n")
m1.pausar()
print("\n")

m1.reproduzir()
print("Quantidade de reproduções:", m1.reproducoes)
```

```python
m2 = Musica("Enemy", "Imagine Dragons", "Musica alternativa", 3, 2021)
print("Antes de atualizar")
print("-"*50)
m2.mostrar_detalhe()
print("\n")

m2.alterar_titulo("Roar")
m2.alterar_cantor("Katy Perry")
m2.alterar_genero("Pop")
m2.alterar_duracao(4)
m2.alterar_ano(2013)

print("\n")
print("Depois de atualizar")
print("-"*50)
m2.mostrar_detalhe()
```

```python
m2.favoritar()
print("\n")
m2.desfavoritar() 
```
